Problem Solving Techniques with Python

1️⃣ Brute Force Approach

```python
def two_sum_brute_force(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):  # Try all pairs
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

# Test case
print(two_sum_brute_force([2, 7, 11, 15], 9))  # Output: [0, 1]
```

2️⃣ Hashing (Using Dictionary / HashMap)

```python
def two_sum(nums, target):
    num_map = {}  # Dictionary for storing numbers and their indices
    for i, num in enumerate(nums):
        complement = target - num  # Find the number needed to reach the target
        if complement in num_map:
            return [num_map[complement], i]  # Return the indices
        num_map[num] = i
    return []

# Test case
print(two_sum([2, 7, 11, 15], 9))  # Output: [0,1]
```

3️⃣ Two Pointers Technique

```python
def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1  # Move left pointer to increase sum
        else:
            right -= 1  # Move right pointer to decrease sum
    return []

# Test case
print(two_sum_sorted([1, 2, 3, 4, 6], 5))  # Output: [0, 3]
```

4️⃣ Sliding Window

```python
def length_of_longest_substring(s):
    char_set = set()
    left = 0
    max_length = 0

    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)

    return max_length

# Test case
print(length_of_longest_substring("abcabcbb"))  # Output: 3
```

5️⃣ Recursion & Backtracking

```python
def generate_parentheses(n):
    result = []

    def backtrack(current, open_count, close_count):
        if len(current) == 2 * n:
            result.append(current)
            return
        if open_count < n:
            backtrack(current + "(", open_count + 1, close_count)
        if close_count < open_count:
            backtrack(current + ")", open_count, close_count + 1)

    backtrack("", 0, 0)
    return result

# Test case
print(generate_parentheses(3))  
# Output: ["((()))", "(()())", "(())()", "()(())", "()()()"]
```

6️⃣ Binary Search (Divide & Conquer)

```python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1

    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

# Test case
print(binary_search([1, 2, 3, 4, 5, 6, 7], 5))  # Output: 4
```

## Summary of Time Complexity for All Techniques and When to Use Each One
| **Technique**                | **Best Case** | **Worst Case** | **Usage Scenario** |
|------------------------------|-------------|-------------|-------------------------------------------|
| **Brute Force**              | O(N)        | O(N²) or O(N³) | When problem size is small and no better approach is obvious |
| **Hashing (Dictionary/Set)**  | O(1)        | O(N)        | Fast lookups, frequency counting, avoiding duplicates |
| **Two Pointers**             | O(N)        | O(N)        | Sorted arrays, finding pairs, merging two lists |
| **Sliding Window**           | O(N)        | O(N)        | Problems involving contiguous subarrays (max sum, longest unique substring) |
| **Recursion & Backtracking** | O(2ⁿ)      | O(2ⁿ)      | Exploring all possible solutions (subsets, permutations, valid parentheses) |
| **Binary Search**            | O(1)        | O(log N)    | Searching in sorted arrays, optimization problems |
| **Dynamic Programming (DP)** | O(N)        | O(N²) or worse | Problems with overlapping subproblems (Fibonacci, knapsack, longest subsequence) |
| **Graph Traversal (BFS & DFS)** | O(V + E)    | O(V + E)    | Pathfinding, connectivity in graphs, shortest paths |

### Understanding Time Complexity Definitions

| **Time Complexity** | **Definition (Beginner Friendly Explanation)** | **Best & Worst Case Scenarios** |
|--------------------|---------------------------------------------|--------------------------------|
| **O(1) - Constant Time** | The operation takes the **same amount of time** regardless of the input size (N). | Best: Always fast ⚡<br>Worst: Still fast ✅ |
| **O(log N) - Logarithmic Time** | The number of operations grows **slowly** as input increases. Each step halves the input. (e.g., Binary Search) | Best: Super efficient 🚀<br>Worst: Slows down gradually 📉 |
| **O(N) - Linear Time** | The time taken increases **proportionally** to input size (e.g., scanning an array once). | Best: Reasonable for large inputs ✔<br>Worst: Can be slow for large N ❗ |
| **O(N log N) - Linearithmic Time** | Used in efficient sorting algorithms (Merge Sort, QuickSort). Faster than O(N²). | Best: Very efficient for sorting 🌟<br>Worst: Still much better than O(N²) ✔ |
| **O(N²) - Quadratic Time** | The number of operations grows **quadratically** (e.g., nested loops over an array). | Best: Okay for small inputs ⚠<br>Worst: Extremely slow for large N ❌ |
| **O(2ⁿ) - Exponential Time** | The time **doubles** with each additional input (e.g., Recursion problems like Fibonacci, Backtracking). | Best: Horrible for large inputs ❌<br>Worst: Becomes infeasible even for N > 30 ❌❌❌ |
| **O(V + E) - Graph Traversal** | The time depends on the number of **vertices (V) and edges (E)** in a graph. Used in BFS/DFS. | Best: Efficient for sparse graphs 🌍<br>Worst: Can be slow for dense graphs ⚠ |
