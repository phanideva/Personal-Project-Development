## Question statement:

### Given a string s, find the length of the longest substring without repeating characters.

Example 1:
Input: s = "abcabcbb";
Output: 3;
Explanation: The answer is "abc", with the length of 3.

Example 2:
Input: s = "bbbbb";
Output: 1;
Explanation: The answer is "b", with the length of 1.

Example 3:
Input: s = "pwwkew";
Output: 3;
Explanation: The answer is "wke", with the length of 3.

Constraints:
0 <= s.length <= 5 * 10^4;
s consists of English letters, digits, symbols, and spaces.

Step 1: Understanding the Problem
1. We are given a string.
2. We need to find the longest substring that does not contain duplicate characters.
3. We return the length of this longest substring.

Step 2: Thinking of a Solution.
There are multiple ways to solve this problem:

1. Brute Force (O(n²)):
 - Generate all substrings.
 - Check if they have unique characters.
 - Keep track of the maximum length found.
 - Issue: This is inefficient for large inputs.
2. Sliding Window with Hash Set (O(n)):
 - Use a two-pointer approach (left and right).
 - Use a hash set to keep track of characters in the current window.
 - If a character is repeated, move the left pointer.
 - Update the max length found.

The sliding window approach is optimal and efficient.

Step 3: Implementing the Sliding Window Approach

```python
def length_of_longest_substring(s: str) -> int:
    char_set = set()
    left = 0  # Left pointer of the window
    max_length = 0  # Maximum substring length

    for right in range(len(s)):
        # If the character is already in the set, remove from the left
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1  # Shrink the window

        # Add the new character and update max length
        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)

    return max_length
```

Step 4: Explanation of the Code
1. Initialize variables:
 - char_set: Stores unique characters in the current window.
 - left: The left boundary of the sliding window.
 - max_length: Tracks the longest valid substring found.
2. Iterate over the string with right pointer:
 - If s[right] is already in char_set, it means we have a duplicate.
 - We move the left pointer forward and remove characters from the set until s[right] is unique.
 - If s[right] is not in char_set, we add it and update max_length.
3. Complexity Analysis:
 - Each character is added and removed from char_set at most once.
 - This results in an O(n) time complexity.

Step 5: Testing the Solution

```python
print(length_of_longest_substring("abcabcbb"))  # Output: 3
print(length_of_longest_substring("bbbbb"))     # Output: 1
print(length_of_longest_substring("pwwkew"))    # Output: 3
print(length_of_longest_substring(""))          # Output: 0
print(length_of_longest_substring("au"))        # Output: 2
```

Step 6: Fixing and Optimizing
 - If s is empty, the function correctly returns 0.
 - Using a hash set ensures O(1) lookups and removals.
 - Alternative Approach: We can use a dictionary to track character indices for a more optimized sliding window with hashmap approach.

Optimized Version Using Dictionary:

```python
def length_of_longest_substring(s: str) -> int:
    char_index = {}  # Dictionary to store the last seen index of characters
    left = 0
    max_length = 0

    for right in range(len(s)):
        if s[right] in char_index and char_index[s[right]] >= left:
            left = char_index[s[right]] + 1  # Move left pointer past the duplicate

        char_index[s[right]] = right  # Update last seen index
        max_length = max(max_length, right - left + 1)

    return max_length
```

Step 7: Explanation of the Optimized Version
1. Why use a dictionary?
 - Instead of using a set, we track the index of each character.
 - If a duplicate is found, we jump the left pointer directly to char_index[s[right]] + 1.
 - This avoids unnecessary removals and makes the process faster.
2. Time Complexity:
 - Still O(n), but fewer operations compared to the set-based approach.

Final Thoughts
 - This problem is a classic sliding window example.
 - The brute force method is too slow.
 - The set-based sliding window is good.
 - The hashmap sliding window is even better for some cases.

### Variations of the Longest Substring Without Repeating Characters problem

Variation 1: Finding the Substring Instead of Just the Length

Solution:
 - We can modify our sliding window approach to keep track of the actual substring:

```python
def longest_substring(s: str) -> str:
    char_index = {}
    left = 0
    max_length = 0
    start_index = 0  # To store the starting index of the longest substring

    for right in range(len(s)):
        if s[right] in char_index and char_index[s[right]] >= left:
            left = char_index[s[right]] + 1  # Move left pointer past the duplicate
        
        char_index[s[right]] = right  # Update last seen index
        
        # Update max_length and start_index if we found a longer substring
        if right - left + 1 > max_length:
            max_length = right - left + 1
            start_index = left

    return s[start_index:start_index + max_length]  # Return the longest substring
```

Test Cases:

```python
print(longest_substring("abcabcbb"))  # Output: "abc"
print(longest_substring("bbbbb"))     # Output: "b"
print(longest_substring("pwwkew"))    # Output: "wke"
print(longest_substring("dvdf"))      # Output: "vdf"
```

Variation 2: Longest Substring With At Most K Distinct Characters
 - Instead of ensuring all characters are unique, we are now allowed to have at most K distinct characters.

 Example:
 Input: s = "eceba", k = 2; 
 Output: 3; 
 Explanation: The answer is "ece" with length 3.

Solution (Sliding Window + HashMap):
- We use a hashmap to count occurrences of characters and adjust the window accordingly.

```python
def longest_substring_k_distinct(s: str, k: int) -> int:
    char_count = {}
    left = 0
    max_length = 0

    for right in range(len(s)):
        char_count[s[right]] = char_count.get(s[right], 0) + 1  # Add character to hashmap

        # If we exceed K distinct characters, move left pointer
        while len(char_count) > k:
            char_count[s[left]] -= 1
            if char_count[s[left]] == 0:
                del char_count[s[left]]  # Remove character if count becomes zero
            left += 1  # Shrink window

        max_length = max(max_length, right - left + 1)  # Update max length

    return max_length
```

Test Cases:

```python
print(longest_substring_k_distinct("eceba", 2))  # Output: 3 ("ece")
print(longest_substring_k_distinct("aa", 1))     # Output: 2 ("aa")
print(longest_substring_k_distinct("aabbcc", 2)) # Output: 4 ("aabb" or "bbcc")
```
