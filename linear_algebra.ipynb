# Linear Algebra

```python
v = [1, 2, 3]
w = [3, 4, 5]
```

```python
# Adding two vectors
def vector_add(v, w):
    """Adds corresponding elements"""
    return [v_i + w_i for v_i, w_i in zip(v, w)]

result = vector_add(v, w)
print(result)
```

```python
# Subtracting two vectors
def vector_subtract(v, w):
    """Subtracts corresponding elements"""
    return [v_i - w_i for v_i, w_i in zip(v, w)]

result = vector_subtract(v, w)
print(result)
```

```python
a = [23, 67, 76]
b = [26, 76, 54]
c = [27,  83, 98]
vectors = [a, b, c]
```

```python
# Sum of two vectors componentwise
def vector_sum(vectors):
    """Sums all corresponding elements"""
    result = vectors[0]
    print("Inputs given:", result)                          # Start with first vector
    for i, vector in enumerate(vectors[1:], start=1):                   # Then loop over the others
        result = vector_add(result, vector)    # And add them to the result
        print(f"Step {i}: Adding {vector} -> Result {i}: {result}")                      # Show what we're adding     
    return result

result = vector_sum(vectors)
print(f"\nFinal result: {result}")
```

```python
# Reducing the list of vectors using vector_add
# Using higher-order functions
from functools import reduce, partial
def vector_sum(vectors):
    return reduce(vector_add, vectors)

vector_sum = partial(reduce, vector_add)
```

```python
# Multiplying vector by a scalar
def scalar_multiply(c, v):
    """c is a number, v is a vector"""
    return [c * v_i for v_i in v]

result = scalar_multiply(5, v)
print(result)
```

```python
def vector_mean(vectors):
    """compute the vector whose ith element is the mean of the ith elements of the input vectors"""
    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))

result = vector_mean(vectors)
print(f"\nMean vector: {result}")
```

```python
# Dot product is the sum of the vector's componentwise products
def dot(v, w):
    """v_1 * w_1 +....+ v_n * w_n"""
    return sum(v_i * w_i for v_i, w_i in zip(v, w))
result = dot(v, w)
print(f"\nDot Product: {result}")
```

```python
# Vector's sum of squares
def sum_of_squares(v):
    """v_1 * v_1 +....+ v_n * v_n"""
    return dot(v, v)
result = sum_of_squares(v)
print(f"\nSum of Squares: {result}")
```
