# Linear Algebra

```python
v = [1, 2, 3]
w = [3, 4, 5]
```

```python
# Adding two vectors
def vector_add(v, w):
    """Adds corresponding elements"""
    return [v_i + w_i for v_i, w_i in zip(v, w)]

result = vector_add(v, w)
print(result)
```

```python
# Subtracting two vectors
def vector_subtract(v, w):
    """Subtracts corresponding elements"""
    return [v_i - w_i for v_i, w_i in zip(v, w)]

result = vector_subtract(v, w)
print(result)
```

```python
a = [23, 67, 76]
b = [26, 76, 54]
c = [27,  83, 98]
vectors = [a, b, c]
```

```python
# Sum of two vectors componentwise
def vector_sum(vectors):
    """Sums all corresponding elements"""
    result = vectors[0]
    print("Inputs given:", result)                          # Start with first vector
    for i, vector in enumerate(vectors[1:], start=1):                   # Then loop over the others
        result = vector_add(result, vector)    # And add them to the result
        print(f"Step {i}: Adding {vector} -> Result {i}: {result}")                      # Show what we're adding     
    return result

result = vector_sum(vectors)
print(f"\nFinal result: {result}")
```

```python
# Reducing the list of vectors using vector_add
# Using higher-order functions
from functools import reduce, partial
def vector_sum(vectors):
    return reduce(vector_add, vectors)

vector_sum = partial(reduce, vector_add)
```

```python
# Multiplying vector by a scalar
def scalar_multiply(c, v):
    """c is a number, v is a vector"""
    return [c * v_i for v_i in v]

result = scalar_multiply(5, v)
print(result)
```

```python
def vector_mean(vectors):
    """compute the vector whose ith element is the mean of the ith elements of the input vectors"""
    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))

result = vector_mean(vectors)
print(f"\nMean vector: {result}")
```

```python
# Dot product is the sum of the vector's componentwise products
def dot(v, w):
    """v_1 * w_1 +....+ v_n * w_n"""
    return sum(v_i * w_i for v_i, w_i in zip(v, w))
result = dot(v, w)
print(f"\nDot Product: {result}")
```

```python
# Vector's sum of squares
def sum_of_squares(v):
    """v_1 * v_1 +....+ v_n * v_n"""
    return dot(v, v)
result = sum_of_squares(v)
print(f"\nSum of Squares: {result}")
```

```python
# Magnitude
import math

def magnitude(v):
    return math.sqrt(sum_of_squares(v))

result = magnitude(v)
print(f"\nMagnitude (or length): {result}")
```

```python
def squared_distance(v, w):
    """(v_1 - w_1) ** 2 + .... + (v_n - w_n) ** 2"""
    return sum_of_squares(vector_subtract(v, w))

result = squared_distance(v, w)
print(f"\nSquared Distance: {result}")
```

```python
def distance(v, w):
    return math.sqrt(squared_distance(v, w))

result = distance(v, w)
print(f"\nDistance: {result}")
```

```python
def distance(v, w):
    return magnitude(vector_subtract(v, w))

result = distance(v, w)
print(f"\nDistance Magnitude: {result}")
```

```python
# Matrices

A = [[1, 2, 3], 
     [4, 5, 6]]

B = [[1, 2], 
     [3, 4], 
     [5, 6]]
```

```python
# Matrix A has len(A) rows and len(A[0]) columns, it's shape is:

def shape(A):
    num_rows = len(A)
    num_cols = len(A[0]) if A else 0 # number of elements in first row
    return num_rows, num_cols

result = shape(A)
print(f"\nShape of A is: {result}")

# Matrix B has len(B) rows and len(B[0]) columns, it's shape is:

def shape(B):
    num_rows = len(B)
    num_cols = len(B[0]) if B else 0 # number of elements in first row
    return num_rows, num_cols

result = shape(B)
print(f"\nShape of B is: {result}")
```

```python
def get_row(A, i):
    return A[i]                      # A[i] is already the ith row

i = 1
result = get_row(A, i)
print(f"\nGet row {i}: {result}")

def get_column(A, j):
    return [A_i[j]            # jth element of row A_i
            for A_i in A]     # for each row A_i

j = 1
result = get_column(A, j)
print(f"\nGet column {j}: {result}")
```

```python
def make_matrix(num_rows, num_cols, entry_fn):
    """returns a num_rows x num_cols matrix whose (i, j)th entry is entry_fn(i, j)"""
    return [[entry_fn(i, j) for j in range(num_cols)] for i in range(num_rows)]

num_rows = 2
num_cols = 3
entry_fn  = lambda i, j: i * j
result = make_matrix(num_rows, num_cols, entry_fn)
print(f"\nMake matrix with rows{num_rows}, columns {num_cols}, with multiplying : {result}")
```

```python
def is_diagonal(i, j):
    return 1 if i == j else 0

identity_matrix = make_matrix(5, 5, is_diagonal)
result = identity_matrix
print(f"\nMake matrix with identifying its diagonal: {result}")
```
