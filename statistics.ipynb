```python
from collections import Counter
import matplotlib.pyplot as plt

# Re-importing necessary library due to code execution environment reset
import random

# Original list of friends
num_friends = [100, 49, 42, 41, 40, 25, 55, 65, 33, 22]

# Generate 100 more random numbers between 10 and 100 to extend the list
random_friends = [random.randint(10, 100) for _ in range(100)]

# Combine the original list with the new random numbers
num_friends = num_friends + random_friends

num_friends
```

```python
friend_counts = Counter(num_friends)
xs = range(101)
ys = [friend_counts[x] for x in xs]
plt.bar(xs, ys)
plt.axis([0, 101, 0, 25])
plt.title("Histogram of Friend Counts")
plt.xlabel("# of friends")
plt.ylabel("# of people")
plt.show()
```
